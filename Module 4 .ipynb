```python
#16.4 Use the sqlite3 module to create a SQLite database called books.db and a table called books with these fields: title (text), author (text), and year (integer).
import sqlite3

with sqlite3.connect("books.db") as conn:
    conn.execute("""
        CREATE TABLE IF NOT EXISTS books (
            title TEXT,
            author TEXT,
            year INTEGER
        )
    """)
```

```python
#16.5 Read books2.csv and insert its data into the book table.
import csv

with open("books2.csv", encoding="utf-8", newline="") as file:
    reader = csv.DictReader(file)
    rows = [
        (row["title"], row["author"], int(row["year"])) 
            for row in reader]
with sqlite3.connect("books.db") as conn:
    conn.executemany(
        "INSERT INTO books (title, author, year) VALUES (?, ?, ?)",
        rows,
    )

    print(f"Added {len(rows)} books.")
```

```python
#16.8 Use the sqlalchemy module to connect to the sqlite3 database books.db that you just made in exercise 16.4. As in 16.6, select and print the title column from the book table in alphabetical order.
from sqlalchemy import create_engine, text

engine = create_engine("sqlite:///books.db")

with engine.connect() as conn:
    titles = conn.execute(text("SELECT title FROM books ORDER BY title"))

    for title in titles.scalars():
        print(title)
```
